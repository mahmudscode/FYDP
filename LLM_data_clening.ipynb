Now, let's import the libraries and set up the paths. We will create a new directory in your Google Drive to save the extracted text files.

```python
import fitz  # PyMuPDF
import pytesseract
from PIL import Image
import os

# Point tesseract to its installation path (often needed in Colab)
# Adjust this path if tesseract is installed in a different location
# For Colab, the default is usually fine, but specifying it can prevent issues.
# pytesseract.pytesseract.tesseract_cmd = r'/usr/bin/tesseract'

# Define the input directory (where PDFs are located)
input_pdf_dir = '/content/drive/MyDrive/pdf_llm'

# Define the output directory (where text files will be saved)
output_text_dir = '/content/drive/MyDrive/pdf_llm_extracted_text'

# Create the output directory if it doesn't exist
os.makedirs(output_text_dir, exist_ok=True)

print(f"PDFs will be read from: {input_pdf_dir}")
print(f"Extracted text files will be saved to: {output_text_dir}")
```

Now, we will iterate through each PDF file in the input directory, perform OCR on each page, and save the extracted text into a new `.txt` file in the output directory.

```python
pdf_files = [f for f in os.listdir(input_pdf_dir) if f.endswith('.pdf')]

if not pdf_files:
    print(f"No PDF files found in {input_pdf_dir}")
else:
    print(f"Found {len(pdf_files)} PDF files. Starting text extraction...")
    for pdf_file in pdf_files:
        pdf_path = os.path.join(input_pdf_dir, pdf_file)
        text_filename = os.path.splitext(pdf_file)[0] + '.txt'
        output_text_path = os.path.join(output_text_dir, text_filename)

        extracted_text = []
        try:
            doc = fitz.open(pdf_path)
            print(f"Processing {pdf_file}...")
            for page_num in range(doc.page_count):
                page = doc.load_page(page_num)

                # Render page as a high-resolution image
                pix = page.get_pixmap(matrix=fitz.Matrix(300/72, 300/72)) # 300 DPI
                img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)

                # Perform OCR on the image
                text_from_page = pytesseract.image_to_string(img, lang='eng+ben') # Assuming English and Bengali content
                extracted_text.append(text_from_page)
            doc.close()

            # Write the extracted text to a .txt file
            with open(output_text_path, 'w', encoding='utf-8') as f:
                f.write('\n'.join(extracted_text))
            print(f"Successfully extracted text from {pdf_file} to {output_text_path}")

        except Exception as e:
            print(f"Error processing {pdf_file}: {e}")
    print("Text extraction complete for all available PDFs.")
```
